```python
import numpy as np
import pandas as pd

# Set seed for reproducibility
np.random.seed(42)

# Create monthly date range
months = pd.date_range(start='2022-01-01', end='2024-12-01', freq='MS')

# Create DataFrame
df_sales = pd.DataFrame({
    'Month': months,
    
    'Electronics': np.random.randint(15000, 45000, len(months)) 
                   + np.linspace(0, 10000, len(months)),
    
    'Clothing': np.random.randint(10000, 30000, len(months)) 
                + np.linspace(0, 5000, len(months)),
    
    'Groceries': np.random.randint(20000, 50000, len(months)) 
                 + np.linspace(0, 8000, len(months)),
    
    'Furniture': np.random.randint(5000, 20000, len(months)) 
                 + np.linspace(0, 3000, len(months))
})

# Add Total column
df_sales['Total'] = df_sales[['Electronics', 'Clothing', 'Groceries', 'Furniture']].sum(axis=1)

# Output
print("Sales Dataset Shape:", df_sales.shape)
print(df_sales.head())
```
