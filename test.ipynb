```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
data = {
    'Date': pd.date_range('2023-01-01', periods=100, freq='D'),
    'Region': np.random.choice(['North', 'South', 'East', 'West'], 100),
    'Product': np.random.choice(['Product A', 'Product B', 'Product C'], 100),
    'Category': np.random.choice(['Electronics', 'Clothing', 'Food'], 100),
    'Sales': np.random.randint(1000, 10000, 100),
    'Quantity': np.random.randint(10, 100, 100),
    'Profit': np.random.randint(100, 2000, 100)
}
df = pd.DataFrame(data)


df.head(10)
```

```python
region_avg = df.groupby('Region')['Sales'].mean()
print(region_avg)
```

```python
region_stats = df.groupby('Region')['Sales'].agg(['mean', 'sum', 'count', 'min', 'max'])
print(region_stats)
```

```python
# Define custom function
def range_calc(x):
    return x.max() - x.min()

# Apply custom function
custom_agg = df.groupby('Region').agg({
    'Sales': [range_calc, 'mean'],
    'Profit': [range_calc, lambda x: x.sum() / x.count()]
})
#print("\nCustom Aggregations:")
custom_agg
```

```python
# Apply multiple aggregation functions
multi_agg = df.groupby('Region').agg({
    'Sales': ['sum', 'mean', 'std'],
    'Quantity': ['sum', 'mean'],
    'Profit': ['sum', 'max', 'min']
})
print("\nMultiple Aggregations:")
multi_agg
```

```python
#if hierarchical is possibe then only is useful
# Create hierarchical grouping
hierarchical = df.groupby(['Region', 'Product', 'Category'])['Sales'].sum()
print("\nHierarchical Grouping:")
print(hierarchical.head(15))

# Convert to DataFrame with unstack
unstacked = hierarchical.unstack(level='Category')
print("\nUnstacked by Category:")
print(unstacked)
```

```python
# Dataset 1: Employee Information
employees = pd.DataFrame({
    'emp_id': [1, 2, 3, 4, 5],
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'department_id': [101, 102, 101, 103, 102],
    'salary': [50000, 60000, 55000, 65000, 58000]
})

# Dataset 2: Department Information
departments = pd.DataFrame({
    'dept_id': [101, 102, 103, 104],
    'dept_name': ['Sales', 'IT', 'HR', 'Finance'],
    'location': ['New York', 'San Francisco', 'Chicago', 'Boston']
})

# Dataset 3: Employee Projects
projects = pd.DataFrame({
    'emp_id': [1, 2, 2, 3, 5],
    'project_name': ['Project A', 'Project B', 'Project C', 'Project A', 'Project D'],
    'hours': [120, 150, 80, 100, 90]
})

print("Employees:")
print(employees)
print("\nDepartments:")
print(departments)
print("\nProjects:")
print(projects)
```

```python
inner_merge = pd.merge(
    employees, 
    departments,
    left_on='department_id',
    right_on='dept_id',
    how='inner'
)

print("INNER MERGE:")
inner_merge
```

```python
right_merge = pd.merge(
    employees,
    departments,
    left_on='department_id',
    right_on='dept_id',
    how='right'
)

print("RIGHT MERGE:")
right_merge
```

```python
outer_merge = pd.merge(
    employees,
    departments,
    left_on='department_id',
    right_on='dept_id',
    how='outer'
)

print("OUTER MERGE:")
outer_merge
```
