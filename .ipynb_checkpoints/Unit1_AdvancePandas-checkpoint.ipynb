### Conditional Formatting

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Aman", "Rahul", "Sneha", "Priya", "Kabir"],
    "Age": [21, 34, 19, 29, 41],
    "Score": [88, 67, 92, 74, 90],
    "City": ["Delhi", "Mumbai", "Delhi", "Chennai", "Mumbai"]
}

df = pd.DataFrame(data)
df
```

```python
df["Age"] > 25
```

```python
# Extract rows where Age > 25
df[df["Age"] > 25]
```

```python
# Multiple Conditions
# Example 1: Age > 25 AND Score > 70
df[(df["Age"] > 25) & (df["Score"] > 70)]
```

```python
#OR
df[(df["Age"]>25) | (df["Score"]>70)]
```

```python
# Not
df[(df['City'] != 'Delhi')]
```

```python
df[~(df['City'] == 'Delhi')]
```

```python
df[(df['City'] == 'Mumbai') | (df['City'] == 'Chennai')]
```

```python
# isin() - for multiple ==
df[df['City'].isin(['Mumbai','Chennai'])]
```

### MULTI-INDEXING

```python
arrays = [
    ['A', 'A', 'B', 'B', 'C', 'C'],
    ['one', 'two', 'one', 'two', 'one', 'two']
]
index = pd.MultiIndex.from_arrays(arrays, names=['First', 'Second'])
df_multi = pd.DataFrame(np.random.randn(6, 2), 
                        index=index, 
                        columns=['Value1', 'Value2'])
```

```python
df_multi
```

```python
# Create regular DataFrame
sales_data = {
    'Region': ['North', 'North', 'South', 'South', 'East', 'East'],
    'Quarter': ['Q1', 'Q2', 'Q1', 'Q2', 'Q1', 'Q2'],
    'Sales': [100, 150, 120, 180, 90, 130],
    'Profit': [20, 30, 25, 35, 15, 28]
}
df_sales = pd.DataFrame(sales_data)

# Set multiple columns as index
df_sales_multi = df_sales.set_index(['Region', 'Quarter'])
print("\nSales MultiIndex:")
df_sales_multi
```

```python
# Create from list of tuples
tuples = [('A', 'x'), ('A', 'y'), ('B', 'x'), ('B', 'y')]
index = pd.MultiIndex.from_tuples(tuples, names=['Level1', 'Level2'])
df_tuples = pd.DataFrame(np.random.randn(4, 2), 
                         index=index, 
                         columns=['Col1', 'Col2'])
print("\nMultiIndex from tuples:")
df_tuples
```

```python
# Create from cartesian product
iterables = [['A', 'B'], ['x', 'y', 'z']]
index = pd.MultiIndex.from_product(iterables, names=['First', 'Second'])
df_product = pd.DataFrame(np.random.randn(6, 2), 
                          index=index, 
                          columns=['Value1', 'Value2'])
print("\nMultiIndex from product:")
df_product
```

# Aggregation

```python
np.random.seed(42)
data = {
    'Date': pd.date_range('2023-01-01', periods=100, freq='D'),
    'Region': np.random.choice(['North', 'South', 'East', 'West'], 100),
    'Product': np.random.choice(['Product A', 'Product B', 'Product C'], 100),
    'Category': np.random.choice(['Electronics', 'Clothing', 'Food'], 100),
    'Sales': np.random.randint(1000, 10000, 100),
    'Quantity': np.random.randint(10, 100, 100),
    'Profit': np.random.randint(100, 2000, 100)
}
df = pd.DataFrame(data)


df.head(10)
```

```python
print(f"\nDataset shape: {df.shape}")
```

### Basic GroupBy Operations
#### Single Column Grouping

```python
# Group by single column and calculate mean
region_avg = df.groupby('Region')['Sales'].mean()
#print("\nAverage Sales by Region:")
print(region_avg)
```

```python
# Multiple aggregations on single column
region_stats = df.groupby('Region')['Sales'].agg(['mean', 'sum', 'count', 'min', 'max'])
#print("\nSales Statistics by Region:")
print(region_stats)
```

#### Multiple Column Grouping

```python
# Group by multiple columns
product_region = df.groupby(['Region', 'Product'])['Sales'].sum()
print("\nTotal Sales by Region and Product:")
print(product_region)
```

```python
# Display as DataFrame instead of Series
product_region_df = df.groupby(['Region', 'Product'])['Sales'].sum().reset_index()
print("\nAs DataFrame:")
product_region_df
```

### Aggregation Functions
#### Using .agg() with Multiple Functions

```python
# Apply multiple aggregation functions
multi_agg = df.groupby('Region').agg({
    'Sales': ['sum', 'mean', 'std'],
    'Quantity': ['sum', 'mean'],
    'Profit': ['sum', 'max', 'min']
})
print("\nMultiple Aggregations:")
multi_agg
```

```python
# Flatten column names
multi_agg.columns = ['_'.join(col).strip() for col in multi_agg.columns.values]
#print("\nWith Flattened Columns:")
multi_agg
```

#### Custom Aggregation Functions

```python
# Define custom function
def range_calc(x):
    return x.max() - x.min()

# Apply custom function
custom_agg = df.groupby('Region').agg({
    'Sales': [range_calc, 'mean'],
    'Profit': [range_calc, lambda x: x.sum() / x.count()]
})
#print("\nCustom Aggregations:")
custom_agg
```

```python
# More readable syntax with named aggregations
named_agg = df.groupby('Region').agg(
    total_sales=('Sales', 'sum'),
    avg_sales=('Sales', 'mean'),
    max_profit=('Profit', 'max'),
    total_quantity=('Quantity', 'sum'),
    count=('Sales', 'count')
)
#print("\nNamed Aggregations:")
named_agg
```

## Advanced GroupBy Techniques
#### Iterating Through Groups

```python
# Iterate through each group
print("\nIterating through groups:")
for name, group in df.groupby('Region'):
    print(f"\n{name}:")
    print(group.head(3))
    print(f"Group size: {len(group)}")
```

```python
# Get a specific group
groups = df.groupby('Region')
north_data = groups.get_group('North')
print("\nNorth Region Data (first 5 rows):")
print(north_data.head())
```

```python
# Create hierarchical grouping
hierarchical = df.groupby(['Region', 'Product', 'Category'])['Sales'].sum()
print("\nHierarchical Grouping:")
print(hierarchical.head(15))

# Convert to DataFrame with unstack
unstacked = hierarchical.unstack(level='Category')
print("\nUnstacked by Category:")
print(unstacked)
```

### Filtering Groups

```python
# Filter groups based on group properties
# Keep only regions with more than 20 sales records
filtered_groups = df.groupby('Region').filter(lambda x: len(x) > 20)
print(f"\nRegions with > 20 records: {filtered_groups['Region'].unique()}")
print(f"Original regions: {df['Region'].unique()}")
```

```python
# Keep groups where average sales > 5000
high_sales_regions = df.groupby('Region').filter(lambda x: x['Sales'].mean() > 5000)
print(f"\nRegions with avg sales > 5000:")
print(high_sales_regions.groupby('Region')['Sales'].mean())
```

#### Transform Method

```python
# Add group statistics back to original DataFrame
# Calculate z-score within each region
df['Sales_Mean_Region'] = df.groupby('Region')['Sales'].transform('mean')
df['Sales_Std_Region'] = df.groupby('Region')['Sales'].transform('std')
df['Sales_ZScore'] = (df['Sales'] - df['Sales_Mean_Region']) / df['Sales_Std_Region']


df[['Region', 'Sales', 'Sales_Mean_Region', 'Sales_ZScore']].head(10)
```

```python
df
```

```python
df.columns
```

```python
# Apply custom function to each group
def top_sales(group):
    return group.nlargest(2, 'Sales')

# Get top 2 sales records per region
top_per_region = df.groupby('Region').apply(top_sales)
print("\nTop 2 Sales per Region:")
print(top_per_region[['Region', 'Product', 'Sales']])
```

### Grouping by Time/Date

```python
# Add month and quarter columns
df['Month'] = df['Date'].dt.month
df['Quarter'] = df['Date'].dt.quarter
df['Year'] = df['Date'].dt.year

# Group by time periods
monthly_sales = df.groupby('Month')['Sales'].sum()
#print("\nMonthly Sales:")
print(monthly_sales)
```

```python
# Using resample for time-based grouping
df_timeseries = df.set_index('Date')
weekly_sales = df_timeseries['Sales'].resample('W').sum()
print("\nWeekly Sales:")
print(weekly_sales.head())
```

```python
# Group by multiple time periods
time_region = df.groupby(['Quarter', 'Region'])['Sales'].sum().unstack()
print("\nQuarterly Sales by Region:")
print(time_region)
```

### Size and Count Operations

```python
# Count occurrences
region_counts = df.groupby('Region').size()
print("\nCount of records per Region:")
print(region_counts)
```

```python
# Count non-null values
product_counts = df.groupby('Product')['Sales'].count()
print("\nNon-null Sales count per Product:")
print(product_counts)
```

```python
# Value counts for multiple columns
value_counts = df.groupby(['Region', 'Product']).size().reset_index(name='Count')
print("\nCombination counts:")
value_counts
```

# PIVOT TABLES
Pivot tables provide a way to summarize and reorganize data, similar to Excel pivot tables.


```python
# Create basic pivot table
pivot_basic = df.pivot_table(
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum'
)
#print("\nBasic Pivot Table (Sales by Region and Product):")

pivot_basic
```

### Multiple Values

```python
# Pivot table with multiple value columns
pivot_multi = df.pivot_table(
    values=['Sales', 'Profit', 'Quantity'],
    index='Region',
    columns='Product',
    aggfunc='sum'
)
print("\nPivot Table with Multiple Values:")
pivot_multi
```

### Multiple Aggregation Functions

```python
# Apply different functions to different columns
pivot_mixed = df.pivot_table(
    values=['Sales', 'Quantity'],
    index='Region',
    columns='Product',
    aggfunc={
        'Sales': ['sum', 'mean'],
        'Quantity': ['sum', 'max']
    }
)
print("\nPivot Table with Multiple Aggregation Functions:")
pivot_mixed
```

### Multiple Index/Column Levels

```python
# Hierarchical pivot table
pivot_hierarchical = df.pivot_table(
    values='Sales',
    index=['Region', 'Category'],
    columns=['Product', 'Quarter'],
    aggfunc='sum',
    fill_value=0
)
print("\nHierarchical Pivot Table:")
pivot_hierarchical.head(10)
```

#### Fill Missing Values

```python
# Handle missing values in pivot table
pivot_filled = df.pivot_table(
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum',
    fill_value=0  # Replace NaN with 0
)
print("\nPivot Table with Filled Values:")
pivot_filled
```

# Advanced Pivot Table Operations
#### Using Custom Aggregation Functions

```python
# Custom aggregation in pivot table
def revenue_range(x):
    return x.max() - x.min()

pivot_custom = df.pivot_table(
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc=[revenue_range, 'mean', 'count']
)
print("\nPivot Table with Custom Function:")
pivot_custom
```

### Observed Parameter (for Categorical Data)

```python
# Convert to categorical
df['Region_Cat'] = pd.Categorical(df['Region'])
df['Product_Cat'] = pd.Categorical(df['Product'])

# Use observed=True to only show existing categories
pivot_observed = df.pivot_table(
    values='Sales',
    index='Region_Cat',
    columns='Product_Cat',
    aggfunc='sum',
    observed=True  # Only show observed category combinations
)
# print("\nPivot Table with Observed Categories:")
pivot_observed
```

### Crosstab (Special Case of Pivot Table)

```python
# Crosstab for frequency tables
crosstab_basic = pd.crosstab(
    df['Region'],
    df['Product']
)
print("\nCrosstab - Frequency Count:")
crosstab_basic
```

```python
# Crosstab with values
crosstab_values = pd.crosstab(
    df['Region'],
    df['Product'],
    values=df['Sales'],
    aggfunc='sum'
)
#print("\nCrosstab with Sales Values:")
crosstab_values
```

```python
# Crosstab with margins and percentages
crosstab_margins = pd.crosstab(
    df['Region'],
    df['Product'],
    margins=True,
    normalize='all'  # Can be 'all', 'index', or 'columns'
)
#print("\nCrosstab with Normalized Values:")
crosstab_margins
```
