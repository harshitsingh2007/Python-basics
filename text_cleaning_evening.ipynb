```python
!pip install nltk pandas

import re
import string
import pandas as pd
import nltk

from nltk.tokenize import word_tokenize, sent_tokenize, TweetTokenizer, RegexpTokenizer
from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer, SnowballStemmer, LancasterStemmer, WordNetLemmatizer

for pkg in ["punkt", "punkt_tab", "stopwords", "wordnet", "omw-1.4",
            "averaged_perceptron_tagger", "averaged_perceptron_tagger_eng"]:
    nltk.download(pkg)
```

```python
text = (
    "Hello!!! I bought 2 phones from https://shop.example.com on 12/05/2024. "
    "The batteries are running faster than I expected, but the cameras were NOT good. "
    "Customers are saying <b>better</b> deals exist... Contact me at test@mail.com #disappointed"
)
print(text)
```

```python
def basic_clean(text):
    text = text.lower()                                   # 1
    text = re.sub(r"<.*?>", " ", text)                    # 2
    text = re.sub(r"http\S+|www\.\S+", " ", text)         # 3
    text = re.sub(r"\S+@\S+", " ", text)                  # 4
    text = re.sub(r"\d+", " ", text)                      # 5
    text = text.translate(str.maketrans("", "", string.punctuation))  # 6
    text = re.sub(r"\s+", " ", text).strip()              # 7
    return text

cleaned = basic_clean(text)
print(cleaned)
```

```python
sentences=sent_tokenize(text)
for i,s in enumerate(sentences,1)
print(i,s)
```

```python
token = word_tokenize(cleaned)
print(token)
```

```python
reg_tok=RegexpTokenizer(r"[a-zA-Z]+")
print(reg_tok.tokenize(cleaned))
```

```python
stop_words = set(stopwords.words("english"))
print(len(stop_words))

no_stop=[t for t in token if t not in stop_words]
print(no_stop)
```

```python
custom_stop=set(stop_words)
#custom_stop.update({"phones","batteries"})
negations = {"no","not","don't","nor","never"}
custom_stop -= negations

no_stop=[t for t in token if t not in custom_stop]
print(no_stop)
```

```python
# Stemming
porter = PorterStemmer()
snowball = SnowballStemmer("english")
lancaster = LancasterStemmer()

words = [
    "running", "run", "ran", "batteries", "better",
    "studying", "study", "corridor", "caring"
]

stem_df = pd.DataFrame({
    'word': words,
    'porter': [porter.stem(w) for w in words],
    'snowball': [snowball.stem(w) for w in words],
    'lancaster': [lancaster.stem(w) for w in words]
})

print(stem_df)
```
